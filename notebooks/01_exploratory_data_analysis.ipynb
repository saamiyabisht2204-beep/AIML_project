# AI-Based Student Stress Level Prediction
## Exploratory Data Analysis (EDA)

This notebook analyzes the **Student Lifestyle Dataset** (OpenML ID: 46752) to uncover patterns, feature correlations, and distributions associated with student stress levels (**Low**, **Moderate**, **High**).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Plotting aesthetics
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

### 1. Load Raw Dataset

```python
df = pd.read_csv("../data/raw/student_lifestyle_dataset.csv")
print(f"Dataset shape: {df.shape}")
df.head()
```

### 2. Summary Statistics & Data Types

```python
df.info()
df.describe().T
```

### 3. Target Distribution: Stress Level

```python
stress_counts = df['Stress_Level'].value_counts()
print(stress_counts)

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Stress_Level', order=['Low', 'Moderate', 'High'], palette='mako')
plt.title("Distribution of Student Stress Levels")
plt.xlabel("Stress Level")
plt.ylabel("Student Count")
plt.show()
```

### 4. Bivariate Analysis: Sleep & Study Hours vs Stress Level

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, x='Stress_Level', y='Sleep_Hours_Per_Day', order=['Low', 'Moderate', 'High'], ax=axes[0], palette='Blues')
axes[0].set_title("Sleep Hours by Stress Level")

sns.boxplot(data=df, x='Stress_Level', y='Study_Hours_Per_Day', order=['Low', 'Moderate', 'High'], ax=axes[1], palette='Oranges')
axes[1].set_title("Study Hours by Stress Level")

plt.tight_layout()
plt.show()
```

### 5. Correlation Heatmap

```python
numerical_cols = [
    'Study_Hours_Per_Day',
    'Extracurricular_Hours_Per_Day',
    'Sleep_Hours_Per_Day',
    'Social_Hours_Per_Day',
    'Physical_Activity_Hours_Per_Day',
    'GPA'
]

plt.figure(figsize=(8, 6))
sns.heatmap(df[numerical_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("Numerical Features Correlation Matrix")
plt.show()
```
